```python
import re
```

```python
text = "This text has -19.564 many floaring +998.345 numbers 999 test 1.234 floating"
```

```python
floatingPointPattern = r"[+-]?\d+\.?\d*"
```

```python
re.search(floatingPointPattern, text)
```

```python
re.findall(floatingPointPattern, text)
```

```python
for m in re.finditer(floatingPointPattern, text):
    print(m)
```

```python
for m in re.finditer(floatingPointPattern, text):
    print(f"{m.span()}  ---->  {m.group()}")
```

```python
re.sub(floatingPointPattern, "<floating-point>", text)
```

```python
floatingPointPattern = r"[+-]?(?P<integer>\d+)\.?(?P<decimal>\d*)"
```

```python
m = re.search(floatingPointPattern, text)
m.groupdict()
```
