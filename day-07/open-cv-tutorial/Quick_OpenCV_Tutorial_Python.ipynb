## 2. Read and Display an Image

`cv2.imread()` loads an image into a NumPy array.

`img.shape` returns:

```text
(height, width, channels)
```

OpenCV normally represents color images as **BGR** rather than RGB.

> **Jupyter note:** `cv2.imshow()` may not work in every notebook environment. If it does not, use `matplotlib` for display.


````python
import cv2

img = cv2.imread("image.jpg")

if img is None:
    print("Could not find image.jpg")
else:
    print("Image shape:", img.shape)

    cv2.imshow("Image", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
````

## 3. Convert to Grayscale

A grayscale image has a single intensity channel instead of three color channels.

Use:

```python
cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

Grayscale images are commonly used for operations such as thresholding, edge detection, and contour detection.


````python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Gray", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 4. Resize an Image

Resize to an exact size:

```python
cv2.resize(img, (width, height))
```

You can also scale an image proportionally using `fx` and `fy`.


````python
# Exact size
small = cv2.resize(img, (640, 480))

cv2.imshow("Resized", small)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Proportional scaling
small2 = cv2.resize(img, None, fx=0.5, fy=0.5)
print("Original:", img.shape)
print("Scaled:", small2.shape)
````

## 5. Crop an Image

Cropping uses NumPy slicing:

```python
img[y1:y2, x1:x2]
```

Remember that **rows are `y`** and **columns are `x`**.

So:

```python
crop = img[100:400, 200:600]
```

means `y = 100..399` and `x = 200..599`.


````python
crop = img[100:400, 200:600]

cv2.imshow("Crop", crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 6. Save an Image

Use `cv2.imwrite()` to save an image.

The file format is usually determined from the filename extension.

Examples:

```python
cv2.imwrite("output.jpg", img)
cv2.imwrite("output.png", img)
```


````python
cv2.imwrite("output.jpg", img)
print("Saved output.jpg")
````

## 7. Blur an Image

Blurring reduces small image variations and noise.

A Gaussian blur is commonly used before operations such as edge detection and thresholding.

```python
cv2.GaussianBlur(image, kernel_size, sigma)
```

A `(5, 5)` kernel is a common starting point.


````python
blur = cv2.GaussianBlur(img, (5, 5), 0)

cv2.imshow("Blur", blur)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 8. Edge Detection

The Canny algorithm detects strong intensity changes that often correspond to object boundaries.

Typical workflow:

```text
Color image
    ↓
Grayscale
    ↓
Canny
    ↓
Edges
```

The two threshold values control which intensity changes are treated as edges.


````python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 100, 200)

cv2.imshow("Edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 9. Thresholding

Thresholding converts a grayscale image into a binary image.

With `THRESH_BINARY`:

- pixels above the threshold become `255`
- pixels below the threshold become `0`

This is useful when the object and background have sufficiently different brightness.


````python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, binary = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

cv2.imshow("Binary", binary)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 10. Draw Shapes

OpenCV can draw rectangles, circles, text, lines, and other shapes directly onto an image.

The common color format is **BGR**.

For example:

```python
(0, 255, 0)
```

represents green.

The final argument is usually the line thickness.


````python
# Make a copy so the original image is unchanged
draw = img.copy()

# Rectangle
cv2.rectangle(draw, (100, 100), (400, 300), (0, 255, 0), 2)

# Circle
cv2.circle(draw, (300, 200), 50, (255, 0, 0), 2)

# Text
cv2.putText(
    draw,
    "Rice Grain",
    (100, 80),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 255, 0),
    2
)

cv2.imshow("Shapes", draw)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 11. Webcam

`cv2.VideoCapture(0)` opens the default camera.

Each iteration:

1. Capture a frame.
2. Check whether the frame was received.
3. Process/display the frame.
4. Check for a key press.

Press **q** to quit.

This basic pattern is fundamental for real-time computer vision.


````python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        print("Could not read camera frame")
        break

    cv2.imshow("Camera", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
````

## 12. Detect Contours

A contour is a boundary around a connected region in a binary image.

Basic workflow:

```text
Image
  ↓
Grayscale
  ↓
Threshold
  ↓
Binary image
  ↓
Contours
  ↓
Bounding boxes
```

`cv2.boundingRect()` gives:

```text
x, y, width, height
```

This is a useful bridge between traditional computer vision and object detection.


````python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)

contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

result = img.copy()

for contour in contours:
    x, y, w, h = cv2.boundingRect(contour)

    cv2.rectangle(
        result,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        2
    )

cv2.imshow("Contours", result)
cv2.waitKey(0)
cv2.destroyAllWindows()

print("Number of contours:", len(contours))
````

## 13. HSV Color Detection

HSV represents color using:

- **H** — Hue
- **S** — Saturation
- **V** — Value/brightness

HSV is often more convenient than BGR for selecting a particular color range.

`cv2.inRange()` creates a mask:

```text
White → selected pixels
Black → everything else
```

The HSV limits below are only an example. They usually need to be adjusted for the actual lighting and object.


````python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower = np.array([0, 100, 100])
upper = np.array([20, 255, 255])

mask = cv2.inRange(hsv, lower, upper)

cv2.imshow("HSV Mask", mask)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

## 14. A Small Complete Example

This example combines several OpenCV concepts:

```text
Image
  ↓
Grayscale
  ↓
Threshold
  ↓
Contours
  ↓
Bounding boxes
```

The size filter removes very small contours.

This is a simple traditional computer-vision object-detection pipeline. It works best when the objects can be separated from the background using thresholding.


````python
import cv2

img = cv2.imread("objects.jpg")

if img is None:
    raise FileNotFoundError("Place objects.jpg in the notebook folder.")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, binary = cv2.threshold(
    gray, 120, 255, cv2.THRESH_BINARY
)

contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

result = img.copy()

for contour in contours:
    x, y, w, h = cv2.boundingRect(contour)

    if w > 20 and h > 20:
        cv2.rectangle(
            result,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

cv2.imshow("Detected Objects", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
````
