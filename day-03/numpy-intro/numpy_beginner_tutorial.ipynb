# NumPy Operations — Beginner Tutorial

A short hands-on introduction to NumPy arrays, indexing, arithmetic, aggregation, reshaping, filtering, and matrix operations.

## 1. Installation and Import

Install NumPy if required:

```bash
pip install numpy
```

Then import it using the conventional alias `np`.

````python
import numpy as np
````

## 2. Creating Arrays

````python
# 1D array from a Python list
a = np.array([10, 20, 30, 40])
print(a)

# 2D array
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matrix)
````

## 3. Important Array Properties

- `shape` — number of elements in each dimension
- `ndim` — number of dimensions
- `size` — total number of elements
- `dtype` — data type of elements

````python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("shape:", a.shape)
print("ndim :", a.ndim)
print("size :", a.size)
print("dtype:", a.dtype)
````

## 4. Arithmetic Operations 
NumPy performs element-wise arithmetic on arrays.

````python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition      :", a + b)
print("Subtraction   :", a - b)
print("Multiplication:", a * b)
print("Division      :", a / b)
````

## 5. Operations with a Scalar

Applying an operation to an array and a single value is an example of **broadcasting**.

````python
a = np.array([10, 20, 30])

print(a + 5)
print(a * 2)
print(a / 10)
````

## 6. Indexing

````python
a = np.array([10, 20, 30, 40])

print(a[0])
print(a[2])
print(a[-1])

# 2D indexing
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(matrix[0, 1])  # 20
print(matrix[1, 2])  # 60
````

## 7. Slicing
The stop index is not included: `array[start:stop]`.

````python
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])
````

## 8. Useful Mathematical Functions

````python
a = np.array([10, 20, 30, 40])

print("sum :", np.sum(a))
print("mean:", np.mean(a))
print("min :", np.min(a))
print("max :", np.max(a))
print("std :", np.std(a))

# Other useful functions include:
# np.median(a), np.var(a)
````

## 9. Aggregation Along an Axis

For a 2D array:
- `axis=0` → operate down the rows → column-wise
- `axis=1` → operate across the columns → row-wise

````python
marks = np.array([
    [80, 70, 90],
    [60, 75, 85],
    [90, 95, 80]
])

print("Overall average:", np.mean(marks))
print("Column averages:", np.mean(marks, axis=0))
print("Row averages   :", np.mean(marks, axis=1))
````

## 10. Reshaping

````python
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)

print(b)
````

## 11. Creating Special Arrays

````python
print("Zeros :", np.zeros(5))
print("Ones  :", np.ones(4))
print("Range :", np.arange(0, 10, 2))
print("Random:", np.random.rand(5))
````

## 12. Boolean Filtering
Boolean masks are especially useful in data science and machine learning.

````python
a = np.array([10, 25, 30, 15, 40])

print("Condition:", a > 20)
print("Filtered :", a[a > 20])
````

## 13. Matrix Operations

````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)
````

## 14. Why NumPy?

A key progression is:

**Python List → NumPy Array → Vectorized Operations → Fast Numerical Computing → Data Science / ML / AI**

Instead of manually looping:

```python
result = []
for x in numbers:
    result.append(x * 2)
```

NumPy allows vectorized operations such as:

```python
result = numbers * 2
```

Vectorization is one of the core ideas behind NumPy.
