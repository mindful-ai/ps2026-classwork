### 1.2 Prime Numbers

```python
n = int(input("Enter a number: "))
prime = True
for i in range(2, n):
    if n % i == 0:
        prime = False
        break
if prime:
    print("{0} is a prime number".format(n))
else:
    print("{0} is not a prime number".format(n))
```

##### Understanding loop-else statements

```python
n = int(input("Enter a number: "))

for i in range(2, n):
    if n % i == 0:
        print("{0} is not a prime number".format(n))
        break
else:
    print("{0} is a prime number".format(n))
```
