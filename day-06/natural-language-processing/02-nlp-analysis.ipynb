### NLP Analyses with NLTK

We’ll use this sentence:

"I saw a man with a telescope."

Why? Because it has ambiguity that lets us show each type of analysis.

#### Lexical Analysis (Word-level processing)

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

sentence = "I saw a man with a telescope."

# Tokenization
tokens = word_tokenize(sentence)
print("Tokens:", tokens)

# POS tagging
pos_tags = pos_tag(tokens)
print("POS Tags:", pos_tags)
```

Lexical analysis focuses only on individual words – their identity and grammatical category (noun, verb, etc.).

#### Syntactic Analysis (Grammar structure / parsing)

```python
# Define a simple grammar
grammar = r"""
  NP: {<DT>?<JJ>*<NN>}   # Noun phrase
  VP: {<VB.*><NP|PP>*}   # Verb phrase
  PP: {<IN><NP>}         # Prepositional phrase
"""

cp = nltk.RegexpParser(grammar)
tree = cp.parse(pos_tags)
print(tree)
tree.draw()  # Opens a visual parse tree window
```

Syntax shows two possible parses here:

"I saw [a man] [with a telescope]" → The man has the telescope.

"I saw [a man with a telescope]" → I used the telescope to see the man.

This illustrates syntactic ambiguity.

#### Semantic Analysis (Meaning representation)

```python
from nltk.corpus import wordnet as wn

# Example: word sense disambiguation for "saw" and "man"
for syn in wn.synsets("saw"):
    print("Saw sense:", syn.definition())

for syn in wn.synsets("man"):
    print("Man sense:", syn.definition())
```

Semantic analysis tries to pick the right meaning:

"saw" could mean "cutting tool" or "perceived visually".

Here, context says it’s perceived visually.

Semantics reduces ambiguity by linking words to concepts in WordNet or a knowledge graph.

#### Pragmatic Analysis (Context & real-world use)

```python
# Contextual interpretation
context1 = "We were birdwatching."  
context2 = "We were at a hardware store."  

print("Sentence:", sentence)
print("Context 1 (birdwatching): Likely meaning -> I used a telescope to see the man/birds.")
print("Context 2 (hardware store): Likely meaning -> I saw a man carrying/purchasing a telescope.")
```

Pragmatics goes beyond literal meaning.

It uses real-world context to decide which syntactic/semantic interpretation is correct.
