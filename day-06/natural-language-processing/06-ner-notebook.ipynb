```python
import nltk

# Download required packages (only first time)
nltk.download("punkt")
nltk.download("averaged_perceptron_tagger")
nltk.download("maxent_ne_chunker")
nltk.download("words")

# Example text
text = "Apple is looking to invest $1 billion in India by 2025, said Tim Cook in New Delhi."

# Tokenize into words
words = nltk.word_tokenize(text)

# Part-of-speech tagging
pos_tags = nltk.pos_tag(words)

# Named entity recognition using NLTK's chunker
ner_tree = nltk.ne_chunk(pos_tags, binary=False)

# Print named entities
print("Named Entities:")
for subtree in ner_tree:
    if hasattr(subtree, "label"):  # Check if it's a named entity
        entity = " ".join([token for token, pos in subtree.leaves()])
        label = subtree.label()
        print(f"{entity} -> {label}")
```

```python
import spacy

# Load pre-trained English model
# install it using -> python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")

# Example text
text = "Apple is looking to invest $1 billion in India by 2025, said Tim Cook in New Delhi."

# Process the text
doc = nlp(text)

# Print named entities
print("Entities, their labels, and positions:")
for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_} (start: {ent.start_char}, end: {ent.end_char})")

```
